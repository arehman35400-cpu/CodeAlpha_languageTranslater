```python
import ipywidgets as widgets
from IPython.display import display
from deep_translator import GoogleTranslator , MyMemoryTranslator

languages = {
    'English': 'en',
    'Spanish': 'es',
    'French': 'fr',
    'German': 'de',
    'Chinese': 'zh-CN',
    'Japanese': 'ja',
    'urdu': 'ur',
    'Arabic': 'ar',
    'Russian': 'ru',
    'Italian': 'it',
}
Source_dropdown = widgets.Dropdown(
    options=list(languages.keys()),
    value=None,
    placeholder='Select Source Language',
    description='Source Language:',
)
Target_dropdown = widgets.Dropdown(
    options=list(languages.keys()),
    value=None,
    placeholder='Select Target Language',
    description='Target Language:',
)
input_text = widgets.Textarea(
    placeholder="Type here..."
)

translate_button = widgets.Button(description="Translate")

output_text = widgets.Textarea(
    description="Result:",
    disabled=True
)

status = widgets.Label(value="")

def translate(btn):
    text = input_text.value.strip()

    if not text:
        status.value = "Please write something!"
        return

    if not Source_dropdown.value:
        status.value = "Please select a source language!"
        return

    if not Target_dropdown.value:
        status.value = "Please select a target language!"
        return

    src = languages[Source_dropdown.value]
    tgt = languages[Target_dropdown.value]

    try:
        output_text.value = GoogleTranslator(
            source=src,
            target=tgt
        ).translate(text)

        status.value = "Done!"

    except Exception as e:
        status.value = f"Error: {str(e)}"
        output_text.value = ""

translate_button.on_click(translate)

display(Source_dropdown, Target_dropdown, input_text, translate_button, output_text, status)
```
